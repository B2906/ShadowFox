# 8. Model Implementation

## 8.1 Introduction to GPT-2

GPT-2 (Generative Pre-trained Transformer 2) is a transformer-based autoregressive language model developed by OpenAI. It generates text by predicting the next most probable token based on the previous sequence of words.

Unlike traditional NLP models, GPT-2 learns language patterns from large amounts of text during pretraining and can generate coherent and context-aware responses without task-specific training.

For this project, the Hugging Face Transformers library is used to load the pretrained GPT-2 model.

## 8.2 Understanding the Hugging Face Pipeline

The Hugging Face `pipeline()` API provides a simple and high-level interface for using pretrained Natural Language Processing (NLP) models.

Instead of manually loading the tokenizer, preprocessing the input, passing it through the model, and decoding the output, the pipeline performs all these operations automatically.

For this project, the **text-generation** pipeline is used along with the **GPT-2** language model.

The pipeline internally performs the following operations:

1. Loads the pretrained GPT-2 model.
2. Loads the GPT-2 tokenizer.
3. Converts the input text into numerical tokens.
4. Performs inference using the transformer architecture.
5. Converts generated tokens back into readable text.

```python
from transformers import pipeline

generator = pipeline(
    task="text-generation",
    model="gpt2"
)

print("GPT-2 loaded successfully!")
```

```python
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

print("Tokenizer loaded successfully!")
```

## 8.3 Understanding the GPT-2 Tokenizer

Before GPT-2 processes text, it must convert human-readable sentences into numerical representations.

The GPT-2 tokenizer performs the following tasks:

- Splits text into tokens.
- Converts tokens into numerical IDs.
- Handles punctuation and unknown words efficiently.
- Converts token IDs back into readable text after generation.

```python
sentence = "Artificial Intelligence is transforming healthcare."

tokens = tokenizer.tokenize(sentence)

print(tokens)
```

### Encoding

Encoding is the process of converting the generated tokens into numerical IDs.

These numerical IDs are the actual inputs provided to the GPT-2 neural network for processing.

### Observation

The tokenizer does not always split text into complete words. Instead, it uses subword tokenization, which helps the model efficiently represent a large vocabulary while handling unknown or rare words.

```python
token_ids = tokenizer.encode(sentence)

print(token_ids)
```

### Observation

Each token is mapped to a unique numerical ID. These IDs are used by the transformer model during text generation and prediction.

### Observation

The tokenizer does not always split text into complete words. Instead, it uses subword tokenization, which helps the model efficiently represent a large vocabulary while handling unknown or rare words.

```python
token_ids = tokenizer.encode(sentence)

print(token_ids)
```

```python
decoded_sentence = tokenizer.decode(token_ids)

print(decoded_sentence)
```

### Observation

The decoded text matches the original input sentence, showing that the tokenizer can accurately reconstruct the text from the token IDs.

## 8.4 Text Generation

After loading the GPT-2 model, the next step is to generate text based on user-provided prompts.

GPT-2 predicts the next most probable token repeatedly until the stopping condition (maximum length or end-of-sequence token) is reached.

The quality of generated text depends on various decoding parameters such as maximum length, temperature, top-k sampling, and nucleus (top-p) sampling.

```python
prompt = "Artificial Intelligence is"

output = generator(
    prompt,
    max_length=50
)

print(output[0]["generated_text"])
```

### Explanation

The generator accepts an input prompt and predicts the next sequence of words.

Here,

- **Prompt** → Starting text
- **GPT-2** → Predicts the next token repeatedly
- **Maximum Length** → Limits the total generated tokens

The output demonstrates GPT-2's capability to generate coherent natural language.

## Important Parameters

### prompt

The initial text provided to GPT-2.

Example:

"Artificial Intelligence is"

---

### max_length

Defines the maximum number of tokens in the generated output.

Smaller values generate shorter responses.

Larger values generate more detailed responses.

---

### num_return_sequences

Specifies how many different outputs GPT-2 should generate.

Example:

1

↓

One output

Example:

5

↓

Five different outputs

---

### do_sample

When set to True, GPT-2 samples from multiple possible next words, producing more diverse outputs.

---

### temperature

Controls randomness.

Low temperature

↓

More focused

High temperature

↓

More creative

---

### top_k

Restricts GPT-2 to choosing from the K most probable next words.

---

### top_p

Uses nucleus sampling.

Instead of choosing a fixed number of words, GPT-2 chooses from the smallest set of words whose cumulative probability exceeds p.

```python
prompt = "Artificial Intelligence is"

outputs = generator(
    prompt,
    max_length=60,
    num_return_sequences=3
)

for i, output in enumerate(outputs, start=1):
    print(f"\nResponse {i}\n")
    print(output["generated_text"])
```

### Observation

Although the prompt remains the same, GPT-2 generates different responses because it predicts multiple possible continuations.

This demonstrates the probabilistic nature of transformer-based language models.

```python
prompt = "Machine Learning"

print("Temperature = 0.2\n")

print(generator(
    prompt,
    max_length=50,
    temperature=0.2,
    do_sample=True
)[0]["generated_text"])

print("\n\nTemperature = 1.2\n")

print(generator(
    prompt,
    max_length=50,
    temperature=1.2,
    do_sample=True
)[0]["generated_text"])
```

### Observation

Lower temperature generates safer and more deterministic text.

Higher temperature increases diversity and creativity but may produce less coherent responses.

## 9.2 Experiment 1: Context Understanding

### Objective

To evaluate whether GPT-2 can generate meaningful text while maintaining contextual consistency based on different prompts.

```python
prompts = [
    "Artificial Intelligence is",
    "The future of healthcare depends on",
    "Education is becoming",
    "Climate change is",
    "Machine Learning can"
]

for prompt in prompts:
    print("=" * 80)
    print("Prompt:", prompt)
    print()

    result = generator(
        prompt,
        max_length=60,
        do_sample=True,
        temperature=0.7
    )

    print(result[0]["generated_text"])
    print("\n")
```

### Observation

GPT-2 generated contextually relevant responses for all prompts.

The model maintained the topic introduced in the prompt and produced coherent continuations. However, some outputs contained repetitive phrases or unsupported factual claims, demonstrating that GPT-2 predicts likely text patterns rather than verifying facts.

## 9.3 Experiment 2: Domain Adaptability

### Objective

To evaluate GPT-2's ability to generate meaningful text across multiple application domains.

```python
domains = {
    "Technology": "Artificial Intelligence is revolutionizing",
    "Healthcare": "Doctors are using AI to",
    "Education": "Online learning platforms are",
    "Finance": "Digital banking has become",
    "Sports": "Modern sports analytics"
}

for domain, prompt in domains.items():
    print("=" * 80)
    print("Domain:", domain)
    print()

    output = generator(
        prompt,
        max_length=60,
        do_sample=True,
        temperature=0.7,
        truncation=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(output[0]["generated_text"])
    print("\n")
```

### Observation

GPT-2 adapted its responses to each domain by using relevant terminology and maintaining the context introduced by the prompt. The responses demonstrate the model's broad pretrained knowledge across multiple fields. However, the quality and factual accuracy vary depending on the topic.

## 9.4 Experiment 3: Prompt Length Analysis

### Objective

To evaluate how the length of the input prompt affects the quality and coherence of GPT-2 generated text.

```python
prompts = {
    "Short Prompt": "AI is",
    "Medium Prompt": "Artificial Intelligence is transforming the healthcare industry by",
    "Long Prompt": """Artificial Intelligence is transforming the healthcare industry by helping doctors analyze medical images, predict diseases, and improve patient care. In the future,"""
}

for prompt_type, prompt in prompts.items():
    print("=" * 80)
    print(prompt_type)
    print("=" * 80)

    result = generator(
        prompt,
        max_length=100,
        do_sample=True,
        temperature=0.7,
        truncation=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(result[0]["generated_text"])
    print("\n")
```

### Observation

Short prompts generally produced broader and more open-ended responses.

Medium prompts resulted in more focused and coherent continuations.

Long prompts provided richer context, allowing GPT-2 to generate more detailed and contextually aligned text.

## 9.5 Experiment 4: Temperature Analysis

### Objective

To analyze how the temperature parameter influences the creativity and randomness of GPT-2 generated text.

```python
prompt = "Machine Learning is"

temperatures = [0.2, 0.7, 1.2]

for temp in temperatures:
    print("=" * 80)
    print(f"Temperature = {temp}")
    print("=" * 80)

    output = generator(
        prompt,
        max_length=60,
        temperature=temp,
        do_sample=True,
        truncation=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(output[0]["generated_text"])
    print("\n")
```

### Observation

Low temperature (0.2) generated more deterministic and focused text.

Medium temperature (0.7) provided a balance between coherence and diversity.

High temperature (1.2) generated more creative responses but occasionally produced less coherent text.

## 9.6 Experiment 5: Top-k Sampling Analysis

### Objective

To analyze the effect of the **top-k sampling** parameter on GPT-2 text generation.

Top-k sampling limits the model to selecting the next token from only the **k most probable tokens**. Lower values generally produce more focused text, while higher values increase diversity.

```python
prompt = "Artificial Intelligence will"

top_k_values = [10, 50, 100]

for k in top_k_values:
    print("=" * 80)
    print(f"Top-k = {k}")
    print("=" * 80)

    output = generator(
        prompt,
        max_length=60,
        do_sample=True,
        top_k=k,
        temperature=0.7,
        truncation=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(output[0]["generated_text"])
    print("\n")
```

### Observation

For smaller top-k values (e.g., 10), GPT-2 selected words from a smaller pool, resulting in more focused and predictable responses.

For larger top-k values (e.g., 100), the generated text became more diverse and creative, though occasionally less coherent.

## 9.7 Experiment 6: Top-p (Nucleus Sampling)

### Objective

To analyze the influence of **top-p (nucleus sampling)** on GPT-2 text generation.

Unlike top-k sampling, top-p dynamically selects the smallest set of tokens whose cumulative probability exceeds the specified threshold.

```python
prompt = "Machine Learning is"

top_p_values = [0.5, 0.8, 0.95]

for p in top_p_values:
    print("=" * 80)
    print(f"Top-p = {p}")
    print("=" * 80)

    output = generator(
        prompt,
        max_new_tokens=80,
        do_sample=True,
        top_p=p,
        temperature=0.7,
        truncation=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(output[0]["generated_text"])
    print("\n")
```

### Observation

Lower top-p values generated more conservative and focused text.

Higher top-p values increased lexical diversity and creativity by allowing GPT-2 to sample from a larger cumulative probability distribution.

# 12. Ethical Considerations and Limitations

Large Language Models (LLMs) such as GPT-2 are powerful tools for natural language generation. However, their deployment requires careful consideration of ethical implications and technical limitations.

## 12.1 Ethical Considerations

### Bias in Generated Content

GPT-2 is trained on large-scale internet text, which may contain social, cultural, or linguistic biases. As a result, the model may unintentionally generate biased or stereotypical content.

### Misinformation

The model predicts statistically likely text rather than verified facts. Consequently, it can generate inaccurate or misleading information if its outputs are not reviewed.

### Responsible AI Usage

GPT-2 should be used responsibly, particularly in domains such as healthcare, finance, and law, where incorrect information may have significant consequences.

### Privacy

Although GPT-2 does not intentionally memorize user-specific information during inference, developers should avoid providing confidential or sensitive data as prompts.

---

## 12.2 Limitations

The experiments conducted in this project revealed several limitations.

- GPT-2 occasionally generates repetitive text.
- Responses may contain factual inaccuracies.
- The model does not possess reasoning abilities comparable to more recent language models.
- Knowledge is limited to the data available during training.
- The quality of generated text strongly depends on prompt design and decoding parameters.

---

## 12.3 Mitigation Strategies

To improve the reliability of GPT-2 outputs, the following practices are recommended:

- Use well-structured prompts.
- Carefully tune decoding parameters such as temperature, Top-k, and Top-p.
- Verify factual information using trusted external sources.
- Include human review before deploying generated content in real-world applications.
